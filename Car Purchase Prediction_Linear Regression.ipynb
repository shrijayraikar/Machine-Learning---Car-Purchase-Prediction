```python
import pandas as pd
import numpy as np
import matplotlib as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
```

```python
Past_Data = pd.read_csv("Linear-Regression-_Training-Data_-.csv")
Past_Data.head()
```

```python
# Define Independent (X) and Dependend (Y)

X = Past_Data[["Age", "Education", "Income", "Vehicle Ownership"]]
Y = Past_Data["Purchase hybrid vehicle"]
```

```python
# Train with Linear Regression

CarPre_ml = LinearRegression()
CarPre_ml.fit(X,Y)

print("Model has been trained under LINEAR REGRESSION sucessfully")
```

```python
New_Data = pd.read_csv("Linear-Regression-Python-_Test-data_.csv")
New_Data.head()
```

```python
New_Data["Predection %"] = ((CarPre_ml.predict(New_Data))*100).round(2)
New_Data.head()
```

```python
New_Data.to_excel("Car Purchase Prediction_LinearRegression.xlsx", index=False)
```
